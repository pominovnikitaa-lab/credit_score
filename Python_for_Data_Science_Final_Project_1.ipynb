```python
def avg_delivery_time(df: pd.DataFrame):
    df[['time_delivery', 'time_order']] = df[['time_delivery', 'time_order']].apply(pd.to_datetime)
    diff = df['time_delivery'] - df['time_order']
    return diff.mean().ceil('s')

def avg_delivery_time_category(df: pd.DataFrame, column: str):
    result = df[column].value_counts().to_frame()
    count_list = []
    for value in df[column].value_counts().index:
        count_list.append(avg_delivery_time(df.query(f'{column} == @value')))
    result['avg_delivery_time'] = count_list
    return result.reset_index

    
avg_delivery_time_category(df_proj, 'district')
```
